# Load IFCB data


```python
import pandas as pd

url = (
    "https://raw.githubusercontent.com/"
    "fish-pace/fish-pace-datasets/main/"
    "datasets/ifcb/data/mvco_objects_per_ml.csv"
)

df = pd.read_csv(url)
print(f'number of rows: {len(df)}')
df.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Parse date (robust)
df["date"] = pd.to_datetime(df["date"])

# Pick the species
spp = "Alexandrium_catenella"

# Basic time series plot
plt.figure(figsize=(10, 4))
plt.plot(df["date"], df[spp], ".", markersize=3)
plt.xlabel("Date")
plt.ylabel("Objects per mL")
plt.title("MVCO IFCB — Alexandrium catenella abundance")
plt.tight_layout()
plt.show()
```
